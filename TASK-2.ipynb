```python
import sqlite3
import pandas as pd
df = pd.read_csv("SuperMarket Analysis.csv")
conn = sqlite3.connect("SuperMarket Analysis.db")
df.to_sql("sales", conn, if_exists="replace", index=False)
query = """
SELECT *
FROM sales
LIMIT 10
"""
result = pd.read_sql(query, conn)
print(result)
```
